```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
r_coord=np.load('rcoord.npy')
```

```python
good_y_true=np.load('good_mase_y_true.npy')*1e19
good_y_pred=np.load('good_mase_y_pred.npy')*1e19
good_y_naive=np.load('good_mase_y_naive.npy')*1e19
bad_y_true=np.load('bad_mase_y_true.npy')*1e19
bad_y_pred=np.load('bad_mase_y_pred.npy')*1e19
bad_y_naive=np.load('bad_mase_y_naive.npy')*1e19
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.plot(r_coord,good_y_true, color="grey", linewidth=3, linestyle='dashed', label="True density")
plt.plot(r_coord,good_y_pred, color="midnightblue", linewidth=4, label="Predicted density")
plt.plot(r_coord,good_y_naive, color="limegreen", linewidth=4, label="Baseline density")
#ax.set_title('Example of a good reconstruction', fontsize=26, pad=15)
ax.set_xlabel('Major radius (m)', fontsize=22, labelpad=10)
ax.set_ylabel('Electron density ($m^{-3}$)', fontsize=22, labelpad=10)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=18)

# Grid and legend
ax.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=20)
# Tight layout for spacing
plt.tight_layout()
plt.savefig("good_recon.png", dpi=600, bbox_inches='tight')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.plot(r_coord,bad_y_true, color="grey", linewidth=3, linestyle='dashed', label="True density")
plt.plot(r_coord,bad_y_pred, color="midnightblue", linewidth=4, label="Predicted density")
plt.plot(r_coord,bad_y_naive, color="limegreen", linewidth=4, label="Baseline density")
#ax.set_title('Example of a bad reconstruction', fontsize=26, pad=15)
ax.set_xlabel('Major radius (m)', fontsize=22, labelpad=10)
ax.set_ylabel('Electron density ($m^{-3}$)', fontsize=22, labelpad=10)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=18)

# Grid and legend
ax.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=20)
# Tight layout for spacing
plt.tight_layout()
plt.savefig("bad_recon.png", dpi=600, bbox_inches='tight')
plt.show()
```

```python
mre_slowmod=np.load('mre_slowmod.npy')
mre_fastmod=np.load('mre_fastmod.npy')
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
plt.plot(r_coord,mre_slowmod*100, color="midnightblue", linewidth=4, label="10Hz averaging")
plt.plot(r_coord,mre_fastmod*100, color="limegreen", linewidth=4, label="no averaging (100kHz)")
ax.set_title('Avg. prediction errors along the profile', fontsize=26, pad=15)
ax.set_xlabel('Major radius (m)', fontsize=22, labelpad=10)
ax.set_ylabel('Mean relative error (%)', fontsize=22, labelpad=10)

# Increase tick label size
ax.tick_params(axis='both', which='major', labelsize=18)

# Grid and legend
ax.grid(True, linestyle='--', alpha=0.6)
ax.set_ylim(0,50)
plt.legend(fontsize=20)
# Tight layout for spacing
plt.tight_layout()
plt.savefig("MRE.png", dpi=600, bbox_inches='tight')
plt.show()
```
