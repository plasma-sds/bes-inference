```python
import numpy as np
import tensorflow as tf
```

```python
model=tf.keras.models.load_model("density_prediction_model_newcuration_v2.keras")
model.export('saved_model_dir')
```

```python
! python -m tf2onnx.convert --saved-model saved_model_dir --output model_v2.onnx
```

```python
import onnxruntime as rt
```

```python
sess = rt.InferenceSession("model_v2.onnx")
```

```python
x = np.zeros((1, 40, 1), dtype=np.float32)

# Run inference
input_name = sess.get_inputs()[0].name
result = sess.run(None, {input_name: x})

print("Output:", result)
```
